# Basic math operations in R

R is a statistical programming language that is used to analyze data.  R has a number of very useful **functions** built in, including those that perform basic math and print values to the screen.  In the code cells below, you will find some examples of using R for basic calculations.  Note that the lines that begin with `#` are called code *comments* - they are not executed by R, they are there to help explain what the code is doing.  Feel free to play around with the code in these cells...execute it and mess around with the values if you'd like.

```r
# define variables a and b
a = 4
b = 5

# define variable c as the sum of a and b
c = a + b

# print the value of c to the screen
print("The value of c is:")
print(c)
```

```r
# compute the square of c
c_sqrd = c^2
print("c squared is:")
print(c_sqrd)
```

```r
# compute the square root of c
c_sqr = sqrt(c)
print("The square root of c is:")
print(c_sqr)
```

R natively supports "vectors", which hold sets of numbers.  The math operators in R work element-wise on vectors.  Here is some example code for messing around with vectors.

```r
# vectors
v1 = c(1, 2, 3)

v2 = c(4, 5, 6)

# add vectors element wise
v3 = v1 + v2

print("v3 is now:")
print(v3)
```

```r
# square a vector elementwise
print(v3^2)
```

```r
# sum the elements in a vector
sum_v3 = sum(v3)

print(sum_v3)
```
