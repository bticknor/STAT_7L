Suppose that you roll two independent fair six-sided dice - $D_1$ and $D_2$.  Let $S = D_1 + D_2$ denote the sum of the two rolls.  Calculate $P(S) = 5$, the probability that the sum of the dice equals 5.  Show your work!  No code is required here.

# [Q1] - (b)

Now we will estimate this probability using a simulation!  In R, the `sample()` function randomly draws values from a vector.  The code below simulates rolling one fair six-sided die once, by randomly drawing from the vector `c(1, 2, 3, 4, 5, 6)`.  If you re-run the code multiple times, you will get different dice rolls.

```r
dice = c(1, 2, 3, 4, 5, 6)

roll = sample(dice, size=1)

print("you rolled a:")
print(roll)
```

We can use this R function to simulate rolling dice a very large number of times.  The code below simulates rolling each of two independent dice `n` times, and stores the dice rolls in vectors `die1` and `die2`.  Execute this code to run the experiment and produce a barplot showing the distribution of the `n` sums of dice rolls.

```r
# number of dice rolls
set.seed(1)

# number of dice rolls
n = 50

die1 = sample(dice, size = n, replace = TRUE)  # sample n dice rolls from the roll vector, with replacement!
die2 = sample(dice, size = n, replace = TRUE)

# sum of the n dice roll pairs
dice_sum   = die1 + die2

# count the number of
sum_counts = table(dice_sum)

# barplot of the sums
barplot(sum_counts,
        col = "steelblue", border = "white",
        xlab = "Sum of two dice", ylab = "Count",
        main = paste("Sums of", n, "rolls of two dice"))
```

```r
# compute the empirical probability of the two dice summing to 5 in n trials

val = "5"

empirical_prob = sum_counts[val] / n

paste("Empirical probability of the sum of two dice equalling", val, "is", empirical_prob)
```
