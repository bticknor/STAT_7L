# [Q4] - Relationship between bill length and mass

The code below constructs a scatter plot of bill length and body mass for all penguins in the dataset, colored by the species of the penguin.  Execute the code to generate the plot.

(a) Describe the association between these variables is it positive or negative?  Is the association in the same direction for each species?

```r
species_cols <- c(Adelie = "darkorange", Chinstrap = "purple", Gentoo = "cyan4")

plot(bill_len ~ body_mass, data = penguins,
     col = species_cols[as.character(penguins$species)], pch = 19,
     xlab = "Body mass (g)", ylab = "Bill length (mm)",
     main = "Bill length vs. body mass by species")
legend("topleft", legend = names(species_cols),
       col = species_cols, pch = 19)
```

    function_name = function(input) {
      # steps to compute the result

function, and `return()` sends the result back out.


```r
grams_to_kg = function(x) {
  kg = x / 1000
  return(kg)
}

grams_to_kg(4500)
```
